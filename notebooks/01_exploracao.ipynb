## Descobertas da exploração

- Base com 25.045 linhas e 23 colunas (geração centralizada, ANEEL/SIGA)
- 27 UFs, sem nulos nem grafias diferentes
- Fase: 91% em operação, o resto em construção ou não iniciada
- Fonte: usar DscOrigemCombustivel (6 categorias limpas)
- Potência veio como texto com vírgula decimal, em kW -> converter para número e MW
- 3 códigos (CodCEG) repetidos, todos hidrelétricas -> verificar se são cópias exatas
- 4 colunas quase vazias (sub-bacia, vigência, geração qualificada) -> descartar
- O Pará concentra 53% das linhas e solar 76% -> investigar tamanho das usinas

- 3 códigos (CodCEG) repetidos: mesma usina com dois tipos de outorga
  (Autorização e Concessão). Decisão: manter uma linha por CodCEG.
- Datas em formato aaaa-mm-dd; base gerada em 2026-10-01

```python
df_limpo = df.copy()
```

```python
df_limpo = df_limpo.drop_duplicates(subset="CodCEG", keep="first")
```

```python
print(len(df), len(df_limpo), df_limpo["CodCEG"].is_unique)
```

```python
colunas = [
    "CodCEG",
    "NomEmpreendimento",
    "SigUFPrincipal",
    "SigTipoGeracao",
    "DscFaseUsina",
    "DscOrigemCombustivel",
    "DatEntradaOperacao",
    "MdaPotenciaOutorgadaKw",
    "MdaPotenciaFiscalizadaKw",

]
```

```python
df_limpo = df_limpo[colunas].copy()
```

```python
df_limpo.shape 
```

```python
df_limpo.head()
```

```python
df_limpo["potencia_outorgada_mw"] = pd.to_numeric(df_limpo["MdaPotenciaOutorgadaKw"].str.replace(",", "."), errors="coerce") / 1000
```

```python
df_limpo["potencia_outorgada_mw"].isna().sum()
```

```python
df_limpo["potencia_fiscalizada_mw"] = df_limpo["MdaPotenciaFiscalizadaKw"] / 1000
```

```python
df_limpo[["potencia_outorgada_mw", "potencia_fiscalizada_mw"]].describe()
```

```python
operacao = df_limpo[df_limpo["DscFaseUsina"] == "Operação"]
```

```python
resumo = operacao.groupby("DscOrigemCombustivel").agg(
    usinas=("CodCEG", "count"),
    potencia_mw=("potencia_fiscalizada_mw", "sum"),
)
resumo["pct_usinas"] = (100 * resumo["usinas"] / resumo["usinas"].sum()).round(1)
resumo["pct_potencia"] = (100 * resumo["potencia_mw"] / resumo["potencia_mw"].sum()).round(1)
resumo.sort_values("potencia_mw", ascending=False)
```

```python
pa = operacao[operacao["SigUFPrincipal"] == "PA"]
pa["SigTipoGeracao"].value_counts()
pa["potencia_fiscalizada_mw"].describe()
```

```python
df_limpo["porte"] = pd.cut(
    df_limpo["potencia_outorgada_mw"],
    bins=[-1, 0.075, 5, 100, float("inf")],
    labels=["Micro", "Pequena", "Média", "Grande"],
)
```

```python
operacao = df_limpo[df_limpo["DscFaseUsina"] == "Operação"]
operacao.groupby("porte", observed=True)["potencia_fiscalizada_mw"].agg(["count", "sum"])
```

```python
df_limpo["data_operacao"] = pd.to_datetime(df_limpo["DatEntradaOperacao"], errors="coerce")
df_limpo["ano_operacao"] = df_limpo["data_operacao"].dt.year
```

```python
df_limpo.groupby("DscFaseUsina")["ano_operacao"].agg(["count", "min", "max"])
```

```python
df_limpo.loc[df_limpo["data_operacao"].dt.year == 1900, "data_operacao"] = pd.NaT
df_limpo["ano_operacao"] = df_limpo["data_operacao"].dt.year
```

```python
df_limpo.groupby("DscFaseUsina")["ano_operacao"].agg(["count", "min", "max"])
```

```python
final = df_limpo.rename(columns={
    "CodCEG": "cod_ceg",
    "NomEmpreendimento": "nome",
    "SigUFPrincipal": "uf",
    "SigTipoGeracao": "tipo_geracao",
    "DscFaseUsina": "fase",
    "DscOrigemCombustivel": "fonte",
})[["cod_ceg", "nome", "uf", "tipo_geracao", "fase", "fonte", "porte",
    "data_operacao", "ano_operacao", "potencia_outorgada_mw", "potencia_fiscalizada_mw"]]

final["ano_operacao"] = final["ano_operacao"].astype("Int64")
final.shape
```

```python
final.to_csv("../dados/usinas_limpo.csv", index=False)
```
